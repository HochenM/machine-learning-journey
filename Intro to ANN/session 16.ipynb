Perceptron (Artificial Neuron)

The Perceptron is the simplest unit of a neural network, inspired by biological neurons.

It receives inputs (X), multiplies them by weights (W), adds a bias (b), and applies an activation function.


Where: y = AF(Σ(X·W) + b)

X (Inputs): Features/data entering the neuron.

W (Weights): Importance of each input.

b (Bias): Adjustable threshold that shifts the decision boundary.

Σ(X·W) + b: Linear combination of inputs.

f(): Activation function (e.g., ReLU, Sigmoid, Tanh).

y: Final output of the neuron.


--------------------------

Perceptron is the basic artificial neuron.

Modern neural networks use specialized units:  (Types of Neurons)

- CNN → Convolution Filters
- RNN → Recurrent Cells
- LSTM → LSTM Cells
- GRU → GRU Cells
- Transformers → Attention Heads / Transformer Blocks

---------------------------

A neural network is formed by connecting many computational units (perceptrons, CNN filters, RNN cells, LSTM cells, or attention blocks) through weighted connections. The type of unit determines the type of neural network.

True name = Artificial Neural Network (ANN)

-----

Fully Connected (Dense): a layer where every neuron connects to all neurons in the next layer.

Dropout: a regularization technique that randomly disables neurons during training to reduce overfitting.

Dropout is not a type of neural network.

Neural networks automatically learn feature representations from data and are more powerful for complex tasks such as computer vision, NLP, and speech recognition, but typically require more data and computational resources.



-----

✅ Classical ANN/MLP architecture = Input Layer → Hidden Layer(s) → Output Layer

✅ Modern architectures still have inputs and outputs, but the hidden part may contain CNN filters, LSTM cells, or attention blocks instead of simple perceptrons.

All are NN with different Architectures


-------

-Multilayer Perceptron (MLP) / Classical ANN

A Multilayer Perceptron (MLP) is a basic type of neural network.

Structure: Input layer → Fully connected hidden layer(s) → Output layer

All connections are dense (fully connected).

Each neuron acts like a simple perceptron (weighted sum + activation).

Used for basic prediction/classification tasks.


-Modern Neural Network Architectures:

Still have input and output layers, but hidden layers are different.

Not considered MLPs because hidden layers are not fully connected.

-Examples:

CNN (Convolutional Neural Network) → uses filters/kernels for images

RNN / LSTM → uses memory for sequences (text, time series)

Transformer → uses attention mechanisms

Key Difference:

MLP = only dense (fully connected) layers

Modern models = specialized hidden structures (CNN, RNN, Attention)

----------------

Deep learning = “neural networks with many Hidden layers that learn features automatically”

-------

Each layer can have its own activation function, but:

Inside one layer, all neurons usually use the same activation function.




🔥 Final takeaway

✔ Different layers → can have different activations

✔ Same layer → usually same activation for all neurons

✔ Output layer → special activation depends on task (Classification (binary) → Sigmoid

Multi-class → Softmax --  Regression → No activation (linear)


❌ Not required to be the same across whole network

------

neural networks are trained using Gradient Descent–based methods, so they do use an optimizer (or solver).

Weights + Biases are the parameters the neural network must learn

----------

| Function | Range   | Best Use                     | Main Problem       |
| -------- | ------- | ---------------------------- | ------------------ |
| ReLU     | [0, ∞)  | Hidden layers                | Dead neurons       |
| Sigmoid  | (0, 1)  | Output (binary)              | Vanishing gradient |
| Tanh     | (-1, 1) | Hidden layers (older models) | Vanishing gradient |


Relu : each X>0 --> X ---- each X<0 --> 0

## Build a perceptron from scratch

```python
import numpy as np
from pandas.core.common import random_state

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])

# initialize weights

w = np.random.randn(2)
b = 0
lr = 0.1
max_iter=20

#relu
def step (z):
    return 1 if z >0 else 0

#training
for epoch in range(max_iter):
    for i in range (len(X)):
        z = np.dot(X[i],w) + b
        y_pred = step(z)

        error = y[i] - y_pred

        w = w+lr*X[i]*error
        b = b+lr*error
print(w,b)


```

## Simple neural network using sklearn

```python
from sklearn.neural_network import MLPClassifier
X = [[0,0],[0,1],[1,0],[1,1]]
y = [0,0,0,1]

model = MLPClassifier(hidden_layer_sizes=(5,20),max_iter=1000)
model.fit(X,y)

print(model.predict(X))


#Input Layer → Hidden Layer 1 (5 neurons) → Hidden Layer 2 20 neurons) → Output Layer
```

## Activation function intuition

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)

relu = np.maximum(0, x)
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)

plt.plot(x, relu, label="ReLU")
plt.plot(x, sigmoid, label="Sigmoid")
plt.plot(x, tanh, label="Tanh")
plt.legend()
plt.show()
```
