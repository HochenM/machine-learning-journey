# Viola–Jones Roadmap & Summary (Save This)

## What You Must Remember

Viola-Jones is a **face detection algorithm** that combines:

```text
1. Haar Features
2. Integral Image
3. AdaBoost
4. Cascade Classifier
```

Pipeline:

```text
Image
 ↓
Sliding Window (24×24)
 ↓
Integral Image
 ↓
Haar Features
 ↓
AdaBoost
 ↓
Cascade Classifier
 ↓
Face / Not Face
```

---

# Step 1: Sliding Window

The algorithm scans the image using a:

```text
24 × 24 window
```

Example:

```text
Whole Image
+----------------------+
|                      |
|   [24×24 Window]     |
|                      |
+----------------------+
```

Move window:

```text
→ → →
↓ ↓ ↓
```

Each window is checked independently.

Question:

```text
Is this 24×24 patch a face?
```

---

# Step 2: Haar Features

A Haar feature measures brightness differences.

Example:

```text
-1 -1
-1 -1
------
+1 +1
+1 +1
```

Meaning:

```text
Bottom brightness
-
Top brightness
```

If bottom is brighter:

```text
Large Positive Value
```

If top is brighter:

```text
Large Negative Value
```

---

## Important

A Haar feature is NOT always:

```text
24×24
```

The:

```text
24×24
```

is the detection window.

Features can be:

```text
2×2
2×4
4×2
6×3
...
```

and are placed everywhere inside the window.

---

# Step 3: Integral Image

Problem:

```text
180,000 Haar Features
```

Need many rectangle sums.

Too slow.

Solution:

```text
Integral Image
```

Stores cumulative sums.

Allows:

```text
Rectangle Sum
```

in:

```text
O(1)
```

time.

This makes Haar computation extremely fast.

---

# Step 4: AdaBoost

Initially:

```text
~180,000 Haar Features
```

Most are useless.

AdaBoost finds:

```text
Best Feature
↓
Next Best Feature
↓
Next Best Feature
```

until only useful features remain.

Example:

```text
180,000 Features
↓
500 Selected Features
```

---

## Weak Classifier

Each selected feature becomes:

```text
Weak Classifier
```

Example:

```text
If Feature17 > Threshold
    Face
Else
    Not Face
```

One selected Haar feature ≈ one weak classifier.

---

## Strong Classifier

AdaBoost combines:

```text
Weak1
Weak2
Weak3
...
Weak500
```

into:

```text
Strong Classifier
```

---

# Step 5: Cascade Classifier

Problem:

```text
100,000 windows
```

Most are:

```text
Not Faces
```

Running 500 weak classifiers on every window is expensive.

---

## Stage 1

Use:

```text
2 weak classifiers
```

Reject obvious non-faces.

Example:

```text
100,000
↓
10,000 survive
```

---

## Stage 2

Use:

```text
5 weak classifiers
```

```text
10,000
↓
1,000 survive
```

---

## Stage 3

Use:

```text
20 weak classifiers
```

```text
1,000
↓
100 survive
```

---

Continue:

```text
Stage 1
↓
Stage 2
↓
Stage 3
↓
...
```

Only the strongest candidates reach the final stage.

---

# Relation Between AdaBoost and Cascade

This is where many students get confused.

Correct flow:

```text
180,000 Haar Features
↓
AdaBoost
↓
500 Selected Weak Classifiers
↓
Split Into Cascade Stages
↓
Face Detector
```

AdaBoost creates/selects the weak classifiers.

Cascade organizes them for fast rejection.

# Interview Summary (10 Seconds)

> Viola-Jones detects faces by scanning a 24×24 window over an image, computing Haar-like features efficiently using an integral image, selecting the most informative features with AdaBoost, and arranging them into a cascade classifier that quickly rejects non-face regions.
